## Ejercicio 1 

Dos dados se lanzan una vez y se observa el total obtenido. Usa una simulación para encontrar la probabilidad estimada de que la puntuación sea mayor a 7 o un número par. Una simulación es una repetición del mismo experimento multiples veces para observar su comportamiento:

- Ejecuta el experimento 1000 veces (lanza 2 dados 1000 veces, y suma el número de ambos dados).
- Lleva ua cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
- Divide el número del paso 2 entre el número de iteraciones(1000).

```python
#importar librerias

import random

lanza_dado = 0  #variable contador

for i in range(1000):    #bucle para dos dados 
     dado1 = random.randint(1, 6)
     dado2 = random.randint(1, 6)
     suma = dado1 + dado2

if suma > 7 or suma % 2 == 0:  # suma las veces que la condicion se cumple
     lanza_dado += 1

# probabilidad
 
probabilidad = lanza_dado / 1000   #en 1000 veces

print("Veces que se cumplió:", lanza_dado)
print("Probabilidad estimada:", probabilidad)
print("En porcentaje:", probabilidad * 100, "%")
 
```

## Ejercicio 2

Una caja contiene 10 bolas blancas, 20 bolas rojas y 30 bolas verdes. Si tomamos 5 bolas de la caja reemplazándolas (tomando una bola, anotando el color y luego regresandola a la caja). Queremos saber la probabilidad de:

1. Tomar 3 blancas y 2 rojas.
2. Toas todas del mismo color.

Ejecuta el experimento 1000 veces y calcula las probabilidades que mencionadas.

```python
ball_box = {}

# Crea la cajas con las bolas
for i in range(60):
    if i < 10:
        ball_box[i] = "White"
    elif (i > 9) and (i < 30):
        ball_box[i] = "Red"
    else:
        ball_box[i] = "Green"

print(ball_box)
```

```python
# Contadores para las dos condiciones
contador_3blancas_2rojas = 0
contador_todas_mismo_color = 0

# Repetir el experimento 1000 veces
for experimento in range(1000):
    
    # Contadores para este experimento
    blancas = 0
    rojas = 0
    verdes = 0
    
    # Sacar 5 bolas con reemplazo
    for _ in range(5):
        # Elegir una posición al azar (0 a 59)
        posicion_aleatoria = random.randint(0, 59)
        # Ver qué color está en esa posición
        color = ball_box[posicion_aleatoria]
        
        # Contar el color
        if color == "White":
            blancas += 1
        elif color == "Red":
            rojas += 1
        else:  # Green
            verdes += 1
    
    # Verificar condición 1: 3 blancas Y 2 rojas
    if blancas == 3 and rojas == 2:
        contador_3blancas_2rojas += 1
    
    # Verificar condición 2: Todas del mismo color
    if blancas == 5 or rojas == 5 or verdes == 5:
        contador_todas_mismo_color += 1

# Calcular probabilidades
prob_1 = contador_3blancas_2rojas / 1000
prob_2 = contador_todas_mismo_color / 1000

# Mostrar resultados
print("=" * 50)
print("RESULTADOS DE LA SIMULACIÓN (1000 experimentos)")
print("=" * 50)

print("\n1. Probabilidad de sacar 3 blancas y 2 rojas:")
print(f"   Veces que ocurrió: {contador_3blancas_2rojas}")
print(f"   Probabilidad: {prob_1}")
print(f"   Porcentaje: {prob_1 * 100:.2f}%")

print("\n2. Probabilidad de que todas sean del mismo color:")
print(f"   Veces que ocurrió: {contador_todas_mismo_color}")
print(f"   Probabilidad: {prob_2}")
print(f"   Porcentaje: {prob_2 * 100:.2f}%")
```
